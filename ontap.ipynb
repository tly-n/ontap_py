```python
class SinhVien:
    def __init__(self, msv, hoten, lop):
        self.msv = msv
        self.hoten = hoten
        self.lop = lop

    def info(self):
        return f"MSV: {self.msv}, Họ tên: {self.hoten}, Lớp: {self.lop}"


class svk18A3(SinhVien):
    def __init__(self, msv, hoten):
        super().__init__(msv, hoten, "K18A3")

    def info(self):
        return f"[SV K18A3] MSV: {self.msv}, Họ tên: {self.hoten}, Lớp: {self.lop}"


# Chương trình chính
sv1 = svk18A3("24174600155", "Nguyễn thùy linh")
print(sv1.info())
```
